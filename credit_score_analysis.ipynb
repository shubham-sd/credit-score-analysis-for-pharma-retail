### RFM Analysis

```python
# last order 
last_order_date=df_credit['invoice_date'].max()
last_order_date
```

```python
# importing lifetimes liabrary
import lifetimes

rfm_data= lifetimes.utils.summary_data_from_transaction_data(df_credit,'retailer_names','invoice_date','bill_amount')
rfm_data.head()
```

```python
# removing T
rfm_data = rfm_data.drop('T', axis=1)
rfm_data.head()
```

```python
rfm_data.shape
```

```python
rfm_data.dtypes
```

```python
# Custom Normalize Function.
def norm_func(i):
    x = (i-i.min()) / (i.max() - i.min())
    return x

# Normalizing the data.
df_norm = norm_func(rfm_data)
df_norm.describe()
```

```python
# scree plot or elbow curve.
TWSS = []
k = list(range(2, 10))
```

```python
# We are checking here twss/inertia for 2, 9.
from sklearn.cluster import KMeans

for i in k:
    kmeans = KMeans(n_clusters = i)
    kmeans.fit(df_norm)
    TWSS.append(kmeans.inertia_)
```

```python
# scree plot
plt.plot(k, TWSS, 'ro-'); plt.xlabel("No_of_Clusters"); plt.ylabel("total_within_ss");
```

##### Elbow is at 3, so taking k=3

```python
# Modeling, taking 3 clusters.
model = KMeans(n_clusters = 3)
model.fit(df_norm)
```

```python
model.labels_ # getting labels assign to each row.
```

```python
# saving clusters in dataframe
score = pd.Series(model.labels_)
rfm_data['cr_score'] = score.values
rfm_data.head()
```

```python
rfm_data['cr_score'].value_counts()
```

```python
# separating dependant and independant variable
x = rfm_data.drop('cr_score', axis=1)
y = rfm_data['cr_score']
```

```python
# Splitting the dataset into training and test set.  
from sklearn.model_selection import train_test_split  
x_train, x_test, y_train, y_test= train_test_split(x, y, test_size= 0.20, random_state=42)
```

### Building Model

#### Random Forest

```python
#Fitting RandomForest classifier to the training set  
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, f1_score

rfclassifier = RandomForestClassifier(n_estimators= 100, criterion="gini")  
```

```python
rfclassifier.fit(x_train, y_train)
```

```python
## Evaluation of Test data (Prediction)
y_pred = rfclassifier.predict(x_test)
```

```python
# test accuracy
accuracy_score(y_pred, y_test)
```

```python
# train accuracy
pred_train = rfclassifier.predict(x_train)
```

```python
accuracy_score(pred_train, y_train)
```

#### SVM

```python
from sklearn import svm

rbf = svm.SVC(kernel='rbf', gamma=0.5, C=0.1).fit(x_train, y_train)
poly = svm.SVC(kernel='poly', degree=3, C=1).fit(x_train, y_train)
```

```python
rbf_pred = rbf.predict(x_test)
poly_pred = poly.predict(x_test)
```

```python
poly_accuracy = accuracy_score(y_test, poly_pred)
poly_f1 = f1_score(y_test, poly_pred, average='weighted')
print('Accuracy (Polynomial Kernel): ', "%.2f" % (poly_accuracy*100))
print('F1 (Polynomial Kernel): ', "%.2f" % (poly_f1*100))
```

```python
rbf_accuracy = accuracy_score(y_test, rbf_pred)
rbf_f1 = f1_score(y_test, rbf_pred, average='weighted')
print('Accuracy (RBF Kernel): ', "%.2f" % (rbf_accuracy*100))
print('F1 (RBF Kernel): ', "%.2f" % (rbf_f1*100))
```

#### KNN

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors = 3).fit(x_train, y_train)
```

```python
test_accuracy = knn.score(x_test, y_test)
test_accuracy
```

```python
train_accuracy = knn.score(x_train, y_train)
train_accuracy
```

#### Naive Bayes

```python
# training a Naive Bayes classifier
from sklearn.naive_bayes import GaussianNB
gnb = GaussianNB().fit(x_train, y_train)
```

```python
test_accuracy_nb = gnb.score(x_test, y_test)
train_accuracy_nb = gnb.score(x_train, y_train)
print(f'The test accuracy is: {test_accuracy_nb} \n The train accuracy is: {train_accuracy_nb}')
```
